```python
import numpy as np
import matplotlib.pyplot as plt
from dtaidistance import dtw
from dtaidistance import dtw_visualisation as dtwvis

# Generate time series data
np.random.seed(0)
n_points = 30
x = np.linspace(0, 6 * np.pi, n_points)
series1 = np.sin(x) 
series2 = np.sin(x) + np.random.normal(0, 0.5, n_points)

# Compute DTW
distance, paths = dtw.warping_paths(series1, series2, window=25, psi=2)
best_path = dtw.best_path(paths)


dtwvis.plot_warpingpaths(series1, series2, paths, best_path)
```
